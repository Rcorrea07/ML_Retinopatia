# Preparar caches de imagens (conta pessoal)

Gera, **uma vez por tamanho de imagem**, os zips das imagens processadas usados
pelos treinos: os datasets externos do experimento (DDR, APTOS, DeepDRiD,
IDRiD), o train e as extras de 2015. Caches de 512 px ou mais (~7 a 11 GB os do
EyePACS) não cabem no Drive do projeto: ficam no Drive pessoal do responsável,
na pasta `ProjetoRetinopatia_caches`. Com eles, cada sessão de treino restaura
as imagens em minutos, em vez de gastar ~1 h de A100 processando.

Como usar:

1. Escolha na primeira célula o experimento cujo tamanho de imagem (e cujos
   datasets externos) serão preparados, por exemplo `R12_b0_768_externos`.
2. Abra no Colab da **conta pessoal**, só com CPU (o Colab grátis basta: em
   30/10/2026 tinha 215 GB de disco livre, e o pior momento, o test de 2015,
   precisa de ~110 GB). Ao montar o Drive, autorize a conta pessoal.
3. Rode tudo e deixe a aba aberta. Com os 2 núcleos do Colab grátis, leva
   de 6 a 8 h a 768 px (os externos, ~18 mil fotos, saem na primeira hora). Um
   cache que já existe não é refeito: se a sessão cair, rode de novo.
4. Depois, uma vez só: compartilhe a pasta `ProjetoRetinopatia_caches` com a
   conta do projeto como **Leitor** e, na conta do projeto, adicione um atalho
   para ela em "Meu Drive" (Compartilhados comigo → botão direito → Organizar →
   Adicionar atalho). Os treinos leem os caches por esse atalho, sem gastar a
   cota do projeto.


```python
import os
import subprocess
import sys

URL_REPOSITORIO = "https://github.com/Rcorrea07/ML_Retinopatia.git"
PASTA_REPOSITORIO = "/content/ML_Retinopatia"
REVISAO_CODIGO = "origin/v4-experimentos"

# O tamanho de imagem vem do experimento (src/retinopatia/experimentos.py)
EXPERIMENTO = "R12_b0_768_externos"
os.environ["RETINOPATIA_EXPERIMENTO"] = EXPERIMENTO

# Só este notebook grava caches no Drive
os.environ["RETINOPATIA_GRAVAR_CACHE"] = "1"

# Caminho do zip do dataset no Drive pessoal (ajuste se estiver em outra pasta)
os.environ["RETINOPATIA_ZIP_DATASET"] = (
    "/content/drive/MyDrive/Datasets/diabetic-retinopathy-detection.zip"
)

if not os.path.exists(PASTA_REPOSITORIO):
    subprocess.run(["git", "clone", URL_REPOSITORIO, PASTA_REPOSITORIO], check=True)

subprocess.run(["git", "-C", PASTA_REPOSITORIO, "fetch", "--tags", "origin"], check=True)
subprocess.run(["git", "-C", PASTA_REPOSITORIO, "checkout", "--detach", REVISAO_CODIGO], check=True)

sys.path.insert(0, os.path.join(PASTA_REPOSITORIO, "src"))
```
